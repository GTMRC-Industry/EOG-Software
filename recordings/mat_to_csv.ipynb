```python
import os
import numpy as np
import pandas as pd
from scipy.io import loadmat

mat_file = "./EOG_fixed.mat"
output_dir = "csv_outputs"
keys = ["UDB1", "UDB2", "UDB3", "UDB4", "UDB5"]

os.makedirs(output_dir, exist_ok=True)

mat = loadmat(mat_file, squeeze_me=True, struct_as_record=False)

def to_1d(x):
    return np.asarray(x).squeeze().reshape(-1)

for key in keys:
    if key not in mat:
        print(f"{key} not found")
        continue

    recs = mat[key]

    # Make sure we always have an iterable of records
    if not isinstance(recs, np.ndarray):
        recs = np.array([recs], dtype=object)
    else:
        recs = recs.flatten()

    all_rows = []

    for rec in recs:
        timestamp = to_1d(rec.MatlabTimestamp)
        data1 = to_1d(rec.Data1)


        min_len = min(len(timestamp), len(data1))

        df_part = pd.DataFrame({
            "timestamp": timestamp[:min_len],
            "data1": data1[:min_len],
        })

        all_rows.append(df_part)

    # Merge everything from this UDB into one CSV
    df_final = pd.concat(all_rows, ignore_index=True)

    out_path = os.path.join(output_dir, f"{key}.csv")
    df_final.to_csv(out_path, index=False)
    print(f"Saved {out_path}")
```
