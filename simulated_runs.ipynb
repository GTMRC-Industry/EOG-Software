```python
import pandas as pd
import numpy as np
import time
```

```python
s1 = pd.read_csv('./measurements/csv/s1.csv')
timestep = np.mean(np.diff(s1['time']))
```

```python
rt = []
def reconstruct(sim, timestep):
    for i in range(sim.shape[0]):
        rt.append(sim['signal'][i])
        time.sleep(timestep)
        print(rt[-1])

reconstruct(s1, timestep)

```
